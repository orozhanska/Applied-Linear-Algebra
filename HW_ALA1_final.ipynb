```python
import numpy as  np
import matplotlib.pyplot as plt

def stretch(X, a, b, plot_dtek = True, show_mx = True, return_mx = False): # a scales the x coordinate, b scales the y coord.
    v = X.copy()
    mx = np.array([[a, 0],
                   [0, b]])

    v_new = mx @ v

    if show_mx == True:
      print("The transformation mtrix", mx)
      print("The transformed matrix", v_new)

    if plot_dtek == True:
      for letter in np.split(v_new, [12, 20, 32], axis=1):
          plt.fill(letter[0], letter[1], color="black")

      plt.axis("equal")
      plt.show()

    if return_mx == True:
      return mx, v_new
    else:
      return v_new

def shear(X, a, b, plot_dtek = True, show_mx = True, return_mx = False): # x пересувається на у*а, у на b*x
    v = X.copy()
    mx = np.array([[1, a],
                   [b, 1]])

    v_new = mx @ v

    if show_mx == True:
      print("The transformation mtrix", mx)
      print("The transformed matrix", v_new)

    if plot_dtek == True:
      for letter in np.split(v_new, [12, 20, 32], axis=1):
          plt.fill(letter[0], letter[1], color="black")

      plt.axis("equal")
      plt.show()

    if return_mx == True:
      return mx, v_new
    else:
      return v_new

def reflection(X, a, b, plot_dtek = True, show_mx = True, return_mx = False):
    v = X.copy()
    mx = (1 / (a**2 + b**2)) * np.array([[a**2 - b**2, 2*a*b],
                   [2*a*b, b**2 - a**2]])

    v_new = mx @ v

    if show_mx == True:
      print("The transformation mtrix", mx)
      print("The transformed matrix", v_new)

    if plot_dtek == True:
      for letter in np.split(v_new, [12, 20, 32], axis=1):
          plt.fill(letter[0], letter[1], color="black")

      plt.axis("equal")
      plt.show()

    if return_mx == True:
      return mx, v_new
    else:
      return v_new

def rotation(X, theta, plot_dtek = True, show_mx = True, return_mx = False):
    v = X.copy()
    mx = np.array([[np.cos(theta), -np.sin(theta)],
                   [np.sin(theta), np.cos(theta)]])

    v_new = mx @ v

    if show_mx == True:
      print("The transformation mtrix", mx)
      print("The transformed matrix", v_new)

    if plot_dtek == True:
      for letter in np.split(v_new, [12, 20, 32], axis=1):
          plt.fill(letter[0], letter[1], color="black")

      plt.axis("equal")
      plt.show()

    if return_mx == True:
      return mx, v_new
    else:
      return v_new
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("balraj98/modelnet40-princeton-3d-object-dataset")

print("Path to dataset files:", path)
```

```python
# the following was provided in a lab doc

def read_off(file):
    if 'OFF' != file.readline().strip():
        raise('Not a valid OFF header')
    n_verts, n_faces, n_dontknow = tuple([int(s) for s in file.readline().strip().split(' ')])
    verts = [[float(s) for s in file.readline().strip().split(' ')] for i_vert in range(n_verts)]
    faces = [[int(s) for s in file.readline().strip().split(' ')][1:] for i_face in range(n_faces)]
    return verts, faces

def visualize_model(verts, faces):
    fig = plt.figure(figsize=(10, 10))
    ax = fig.add_subplot(111, projection='3d')

    mesh = Poly3DCollection([[verts[i] for i in face] for face in faces])
    mesh.set_facecolor('cyan')
    mesh.set_edgecolor('black')
    mesh.set_alpha(0.5)

    ax.add_collection3d(mesh)


    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')


    plt.show()
```

```python
def rotate_xy (verts, faces, theta, show_mx = True, plot_img = True, return_mx = False):

  v = np.array(verts).T.copy()

  mx  = np.array([[np.cos(theta), -np.sin(theta), 0],
                  [np.sin(theta), np.cos(theta), 0],
                  [0,0,1]])
  v_new = mx @ v

  if show_mx == True:
    print("The transformation mtrix", mx)
    print("The transformed matrix", v_new)

  if plot_img == True:
    visualize_model(v_new.T, faces)

  if return_mx == True:
    return mx, v_new.T
  else:
    return v_new.T

def rotate_yz(verts, faces, theta, show_mx=True, plot_img=True, return_mx = False):

    v = np.array(verts).T.copy()

    mx = np.array([[1, 0, 0],
                   [0, np.cos(theta), -np.sin(theta)],
                   [0, np.sin(theta), np.cos(theta)]])

    v_new = mx @ v

    if show_mx:
        print("The transformation matrix", mx)
        print("The transformed matrix", v_new)

    if plot_img:
        visualize_model(v_new.T, faces)

    if return_mx == True:
      return mx, v_new.T
    else:
      return v_new.T


def rotate_xz(verts, faces, theta, show_mx=True, plot_img=True, return_mx = False):

    v = np.array(verts).T.copy()

    mx = np.array([[np.cos(theta), 0, -np.sin(theta)],
                   [0, 1, 0],
                   [np.sin(theta), 0, np.cos(theta)]])

    v_new = mx @ v

    if show_mx:
        print("The transformation matrix", mx)
        print("The transformed matrix", v_new)

    if plot_img:
        visualize_model(v_new.T, faces)

    if return_mx == True:
      return mx, v_new.T
    else:
      return v_new.T
```

### Theoretical Questions

1. 0.25pts What are linear transformations?

Linear transformations are linear mappings between two vector mappings of same dimension.

2. 0.25pts What is a linear transformation matrix and how can it be interpreted?

A linear transformation matrix represents a linear mapping between vector spaces. Its columns contain the coordinates of the images of the domain's basis vectors, expressed in the codomain's basis.
Multiplying this matrix by a vector's coordinate column gives the coordinates of its transformed image.

3. 0.25pts What features and properties does a rotation matrix have?

A rotation matrix rotates vectors around the origin (or around an axis in 3D) by a specified angle.

Its main properties are:
- preserves lengths and angles.
- It is orthogonal: R^T*R=I.
- Its determinant equals 1.
- Its inverse equals its transpose: R^{-1}=R^T.
